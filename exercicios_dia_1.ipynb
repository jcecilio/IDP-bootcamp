## Parte 5 — Desafios

### Exercício 18 — Quem vendeu mais?

Descubra qual vendedor vendeu a maior quantidade de produtos.

---



```
df.groupby('vendedor')['quantidade'].sum().sort_values(ascending=False)[0:1]
```

### Exercício 19 — Qual cidade vendeu mais?

Descubra qual cidade teve a maior quantidade de produtos vendidos.



```
#df.groupby('cidade')['quantidade'].sum().sort_values(ascending=False)[0:1]
df.groupby('cidade')['quantidade'].sum().nlargest(1)
```

---

### Exercício 20 — Analisando Recife

Considere somente as vendas realizadas em **Recife**.

Depois, descubra qual produto teve a maior quantidade vendida em Recife.

---


```
#df.groupby('cidade')['quantidade'].sum()
vendas_recife = df.loc[df['cidade'] == 'Recife']

vendas_recife.groupby('produto')['quantidade'].sum().nlargest(1)
```
